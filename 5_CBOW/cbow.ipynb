```python
import nltk
import sys
%pip install gensim nltk
print(sys.executable)
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec
```

```python
# Download tokenizer resources
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# a. Data preparation: read text file
with open("data.txt", "r", encoding="utf-8") as file:
    text = file.read().lower()
```

```python
sentences = [
    word_tokenize(line)
    for line in text.splitlines()
    if line.strip()
]
```

```python
# b. Generate training data
# Word2Vec generates context-target training examples internally

# c. Train CBOW model
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    epochs=100
)

```

```python
# d. Output
print("Vocabulary:", list(model.wv.index_to_key))

word = "learning"
print("\nWord vector for", word, ":")
print(model.wv[word])

print("\nWords similar to", word, ":")
print(model.wv.most_similar(word, topn=3))
```
