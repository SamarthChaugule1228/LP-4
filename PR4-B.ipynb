```python
# 4B. Implement the Continuous Bag of Words (CBOW) Model for the twitter dataset 
# (download from Kaggle) using the below steps: 
# a. Data Collection and Preprocessing: Use Kaggle Twitter dataset or any dataset 
# with tweets. Tokenize the tweets, clean the text, and prepare it for training. 
# b. Train Word2Vec CBOW Model 
# c. Evaluate Results: After training, we’ll check the learned word embeddings and 
# possibly evaluate using some examples.
```

```python
import re
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from gensim.models import Word2Vec

# Download required NLTK data
nltk.download("punkt")
nltk.download("stopwords")

# -------------------------------------------------
# 1. Load dataset
# -------------------------------------------------
df = pd.read_csv(r"C:\BE\LP4\LP IV Updated Datasets\CBOW\twitter_dataset.csv")

# Check column name
print(df.columns)
print(df.head())

# Use the Text column
text_data = df["Text"].fillna("").astype(str).tolist()

# -------------------------------------------------
# 2. Preprocess tweets
# -------------------------------------------------
stop_words = set(stopwords.words("english"))
cleaned_sentences = []

for tweet in text_data:
    tweet = tweet.lower()
    tweet = re.sub(r"http\S+|www\.\S+", "", tweet)      # remove URLs
    tweet = re.sub(r"@\w+|#\w+", "", tweet)             # remove mentions/hashtags
    tweet = re.sub(r"[^a-zA-Z\s]", " ", tweet)          # keep only letters
    tokens = word_tokenize(tweet)
    tokens = [w for w in tokens if w.isalpha() and w not in stop_words and len(w) > 1]
    
    if len(tokens) > 0:
        cleaned_sentences.append(tokens)

print("Total cleaned tweets:", len(cleaned_sentences))
print("Example:", cleaned_sentences[0])

# -------------------------------------------------
# 3. Train CBOW Word2Vec model
# -------------------------------------------------
model = Word2Vec(
    sentences=cleaned_sentences,
    vector_size=100,   # embedding size
    window=3,         # context window
    min_count=1,      # keep words that appear at least once
    sg=0,             # sg=0 => CBOW
    workers=2
)

# Save the model
model.save("twitter_cbow_model.model")

# -------------------------------------------------
# 4. Evaluate / check learned embeddings
# -------------------------------------------------
sample_words = ["nice", "party", "happy", "music", "love"]

for word in sample_words:
    if word in model.wv:
        print(f"\nSimilar words to '{word}':")
        print(model.wv.most_similar(word, topn=5))
    else:
        print(f"'{word}' not in vocabulary")

# Show vector for a word
if "nice" in model.wv:
    print("\nVector length for 'nice':", len(model.wv["nice"]))
    print("First 10 values:", model.wv["nice"][:10])
```
