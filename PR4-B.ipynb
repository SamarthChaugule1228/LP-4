```python
# 4B. Implement the Continuous Bag of Words (CBOW) Model for the twitter dataset 
# (download from Kaggle) using the below steps: 
# a. Data Collection and Preprocessing: Use Kaggle Twitter dataset or any dataset 
# with tweets. Tokenize the tweets, clean the text, and prepare it for training. 
# b. Train Word2Vec CBOW Model 
# c. Evaluate Results: After training, we’ll check the learned word embeddings and 
# possibly evaluate using some examples.
```

```python

# Step 1: Import Libraries

import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec

# Step 2: Load Twitter Dataset

data = pd.read_csv("./LP IV Updated Datasets/CBOW/twitter_dataset.csv")

print(data.head())
print(data.columns)

# Step 3: Preprocessing Data

# Select tweet column (change column name if required)
tweets = data["tweet"].astype(str)

# Convert to lowercase and clean text

tweets = tweets.apply(lambda x: x.lower())

tweets = tweets.apply(
    lambda x: re.sub(r'[^a-z\s]', '', x)
)

# Tokenize tweets

sentences = tweets.apply(lambda x: x.split()).tolist()

print("Total Tweets:", len(sentences))
print("Sample Tweet:", sentences[0])

# Step 4: Train CBOW Word2Vec Model

cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=5,
    min_count=2,
    sg=0,
    workers=4,
    epochs=10
)

# Step 5: Display Word Embedding

word = "good"

if word in cbow_model.wv:
    print("Word Vector:")
    print(cbow_model.wv[word])

# Step 6: Find Similar Words

if word in cbow_model.wv:
    print("Similar Words:")
    print(cbow_model.wv.most_similar(word))

# Step 7: Calculate Similarity

word1 = "good"
word2 = "great"

if word1 in cbow_model.wv and word2 in cbow_model.wv:
    similarity = cbow_model.wv.similarity(word1, word2)
    print("Similarity:", similarity)

# Step 8: Save Model

cbow_model.save("twitter_cbow.model")

print("CBOW Model Trained Successfully!")
```
