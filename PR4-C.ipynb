```python

# Step 1: Import Libraries

import re
from gensim.models import Word2Vec

# Step 2: Load Textual Document

file = open("document3.txt", "r", encoding="utf-8")

text = file.read()

file.close()

print("Original Text:")
print(text)

# Step 3: Preprocessing Data

text = text.lower()

text = re.sub(r'[^a-z\s]', '', text)

# Tokenize Text

words = text.split()

sentences = [words]

print("Tokenized Words:")
print(words)

# Step 4: Train CBOW Word2Vec Model

cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    workers=4,
    epochs=100
)

print("CBOW Model Trained Successfully!")

# Step 5: Display Word Embedding

word = words[0]

print("Word:", word)

print("Word Vector:")
print(cbow_model.wv[word])

# Step 6: Find Similar Words

print("Similar Words:")

print(cbow_model.wv.most_similar(word, topn=3))

# Step 7: Calculate Similarity

if len(words) >= 2:

    word1 = words[0]
    word2 = words[1]

    similarity = cbow_model.wv.similarity(word1, word2)

    print("Similarity:", similarity)

# Step 8: Save Model

cbow_model.save("document_cbow.model")

print("Model Saved Successfully!")
```
