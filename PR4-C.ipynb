```python
import re
from gensim.models import Word2Vec

# Use the document variable already defined in the notebook.
_candidates = ("text_document_3", "document3", "doc3", "text3")
raw_text = next((globals()[name] for name in _candidates if name in globals()), None)

if raw_text is None:
    raise NameError(
        "Define the third document as text_document_3 (or document3, doc3, or text3) first."
    )


# a. Prepare sentences and lowercase word tokens.
if isinstance(raw_text, str):
    sentences = re.split(r"[.!?]+", raw_text)
else:
    sentences = list(raw_text)

tokenized_sentences = [
    re.findall(r"\b\w+\b", sentence.lower())
    for sentence in sentences
]
tokenized_sentences = [tokens for tokens in tokenized_sentences if tokens]

if not tokenized_sentences:
    raise ValueError("Document 3 contains no usable words.")

# b. Train CBOW (sg=0) Word2Vec model.
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=2,
    min_count=1,
    sg=0,
    workers=1,
    seed=42,
    epochs=100,
)

# c. Inspect learned embeddings and evaluate example similarities.
print("Vocabulary:", model.wv.index_to_key)
word = model.wv.index_to_key[0]
print(f"\nEmbedding for '{word}':\n", model.wv[word])

if len(model.wv) > 1:
    print(f"\nWords most similar to '{word}':")
    print(model.wv.most_similar(word, topn=min(5, len(model.wv) - 1)))

# Compare two words if both are in the vocabulary.
example_words = model.wv.index_to_key[:2]
if len(example_words) == 2:
    print(
        f"\nSimilarity between '{example_words[0]}' and '{example_words[1]}':",
        model.wv.similarity(*example_words),
    )
```
